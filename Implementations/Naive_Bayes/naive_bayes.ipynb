# Naive Bayes Classifier


## Import Libraries


```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits, load_iris
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import BernoulliNB, GaussianNB, MultinomialNB
```

## Gaussian Naive Bayes


```python
iris = load_iris()
X_g, y_g = iris.data, iris.target
X_train_g, X_test_g, y_train_g, y_test_g = train_test_split(X_g, y_g, test_size=0.2, random_state=42, stratify=y_g)
gaussian_model = GaussianNB()
gaussian_model.fit(X_train_g, y_train_g)
y_pred_g = gaussian_model.predict(X_test_g)
acc_g = accuracy_score(y_test_g, y_pred_g)
print(f"GaussianNB Accuracy: {acc_g:.4f}\n")
print("GaussianNB Classification Report:")
print(classification_report(y_test_g, y_pred_g, target_names=iris.target_names))
```

## Multinomial Naive Bayes


```python
digits = load_digits()
X_m, y_m = digits.data, digits.target
X_train_m, X_test_m, y_train_m, y_test_m = train_test_split(X_m, y_m, test_size=0.2, random_state=42, stratify=y_m)
multinomial_model = MultinomialNB()
multinomial_model.fit(X_train_m, y_train_m)
y_pred_m = multinomial_model.predict(X_test_m)
acc_m = accuracy_score(y_test_m, y_pred_m)
print(f"MultinomialNB Accuracy: {acc_m:.4f}\n")
print("MultinomialNB Classification Report:")
print(classification_report(y_test_m, y_pred_m, digits=4))
```

## Bernoulli Naive Bayes


```python
X_b = digits.data
y_b = digits.target
X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(X_b, y_b, test_size=0.2, random_state=42, stratify=y_b)
bernoulli_model = BernoulliNB(binarize=7.5)
bernoulli_model.fit(X_train_b, y_train_b)
y_pred_b = bernoulli_model.predict(X_test_b)
acc_b = accuracy_score(y_test_b, y_pred_b)
print(f"BernoulliNB Accuracy: {acc_b:.4f}\n")
print("BernoulliNB Classification Report:")
print(classification_report(y_test_b, y_pred_b, digits=4))
```

## Visualization


```python
model_names = ["GaussianNB\n(Continuous)", "MultinomialNB\n(Counts)", "BernoulliNB\n(Binary)"]
accuracies = [acc_g, acc_m, acc_b]
plt.figure(figsize=(8, 5))
bars = plt.bar(model_names, accuracies, color=["royalblue", "darkorange", "seagreen"], width=0.5)
plt.ylabel("Accuracy")
plt.title("Naive Bayes Variants Performance Comparison")
plt.ylim(0, 1.1)
for bar, acc in zip(bars, accuracies):
    plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.02, f"{acc:.4f}", ha="center", fontweight="bold")
plt.show()
```
