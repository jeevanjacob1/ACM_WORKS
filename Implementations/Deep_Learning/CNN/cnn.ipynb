# Convolutional Neural Network (CNN)

A beginner-friendly implementation of a Convolutional Neural Network (CNN) for handwritten digit classification using the MNIST dataset with TensorFlow and Keras.


## Import Libraries


```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models, datasets
```

## Load Dataset


```python
(X_train, y_train), (X_test, y_test) = datasets.mnist.load_data()

print("Original Training Data Shape:", X_train.shape)
print("Training Labels Shape:       ", y_train.shape)
print("Original Testing Data Shape: ", X_test.shape)
print("Testing Labels Shape:        ", y_test.shape)
```

## Data Preprocessing


```python
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

X_train = X_train.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)

print("Reshaped Training Shape:", X_train.shape)
print("Reshaped Testing Shape: ", X_test.shape)
```

## Model Architecture


```python
model = models.Sequential([
    layers.Conv2D(filters=32, kernel_size=(3, 3), activation="relu", input_shape=(28, 28, 1)),
    layers.MaxPooling2D(pool_size=(2, 2)),
    layers.Conv2D(filters=64, kernel_size=(3, 3), activation="relu"),
    layers.MaxPooling2D(pool_size=(2, 2)),
    layers.Flatten(),
    layers.Dense(units=64, activation="relu"),
    layers.Dense(units=10, activation="softmax")
])

model.summary()
```

## Model Compilation


```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

## Model Training


```python
history = model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.1
)
```

## Model Evaluation


```python
test_loss, test_accuracy = model.evaluate(X_test, y_test, verbose=0)
print(f"Test Loss:     {test_loss:.4f}")
print(f"Test Accuracy: {test_accuracy * 100:.2f}%")
```

## Prediction and Visualization


```python
sample_images = X_test[:5]
sample_labels = y_test[:5]

predictions = model.predict(sample_images)
predicted_classes = np.argmax(predictions, axis=1)

print("Actual Labels:   ", sample_labels)
print("Predicted Labels:", predicted_classes)

plt.figure(figsize=(10, 3))
for i in range(5):
    plt.subplot(1, 5, i + 1)
    plt.imshow(sample_images[i].reshape(28, 28), cmap="gray")
    plt.title(f"Actual: {sample_labels[i]}\nPred: {predicted_classes[i]}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

## CNN Concepts and Workflow Explanation

### Key Components

- **Conv2D**: Applies 2D convolutional filters across the input image to extract local spatial features such as edges, curves, and textures.
- **ReLU**: Rectified Linear Unit activation function $f(x) = \max(0, x)$, introducing non-linearity to capture complex patterns without vanishing gradients for positive inputs.
- **MaxPooling2D**: Downsamples feature maps by taking the maximum value in a window (e.g., $2 \times 2$), reducing spatial dimensions, computational cost, and providing translation invariance.
- **Flatten**: Flattens multidimensional feature maps into a 1D vector so dense layers can process them.
- **Dense**: Fully connected layer that combines features extracted by the convolutional layers to perform classification.
- **Softmax**: Converts final output logits into a probability distribution summing to 1 across the 10 digit classes ($0-9$).
- **Sparse Categorical Crossentropy**: Loss function for multi-class classification when target labels are integers rather than one-hot encoded vectors.
- **Adam Optimizer**: Adaptive Moment Estimation optimization algorithm combining AdaGrad and RMSProp for adaptive parameter learning rates.

### CNN Workflow

1. **Input Image**: Grayscale digit image of size $28 \times 28 \times 1$.
2. **Feature Extraction**: Successive Conv2D and MaxPooling layers learn low-level to high-level spatial patterns.
3. **Vectorization**: Flatten layer converts feature maps into a 1D vector.
4. **Classification**: Dense layers map features to 10 class probabilities.
5. **Output**: Argmax of Softmax probabilities yields the predicted digit class.
